```python
def get_top_artists_by_country(country_code):
    # Get top artists from a country (Note: Spotify doesn't directly support fetching top artists by country)
    # This example uses a search query for top artists in a specific market
    results = sp.search(q='*', type='artist', market=country_code, limit=10)
    artists = [artist['id'] for artist in results['artists']['items']]  # Get artist IDs
    return artists

def get_artist_genres(artist_id):
    # Get genres for a given artist
    artist = sp.artist(artist_id)
    return artist['genres']

# Example: Get top artists for the United States (US)
us_top_artists = get_top_artists_by_country('US')

# Fetch and print genres for top artists
for artist_id in us_top_artists:
    genres = get_artist_genres(artist_id)
    print(f"Artist ID: {artist_id}, Genres: {genres}")
```

```python
def get_top_tracks_in_lithuania():
    # Get top tracks in Lithuania (LT) market
    results = sp.search(q='*', type='track', market='LT', limit=10)
    top_tracks = []
    for track in results['tracks']['items']:
        top_tracks.append({'name': track['name'], 'artist': track['artists'][0]['name'], 'popularity': track['popularity']})
    return top_tracks

def get_top_artists_in_lithuania():
    # Get top artists in Lithuania (LT) market
    results = sp.search(q='*', type='artist', market='LT', limit=10)
    top_artists = []
    for artist in results['artists']['items']:
        top_artists.append({'name': artist['name'], 'followers': artist['followers']['total'], 'popularity': artist['popularity']})
    return top_artists

# Fetch top tracks and artists in Lithuania
lithuania_top_tracks = get_top_tracks_in_lithuania()
lithuania_top_artists = get_top_artists_in_lithuania()

# Print top tracks and artists
print("Top Tracks in Lithuania:")
for track in lithuania_top_tracks:
    print(f"Track: {track['name']}, Artist: {track['artist']}, Popularity: {track['popularity']}")

print("\nTop Artists in Lithuania:")
for artist in lithuania_top_artists:
    print(f"Artist: {artist['name']}, Followers: {artist['followers']}, Popularity: {artist['popularity']}")
```
